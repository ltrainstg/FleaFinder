They somehow got the encryption key and decoded the JSON file in there github, not sure if they where the first but I copied there functions to look at the JSON data below. The [silksong-flea](https://mikkerlo.github.io/silksong-flea/) app does the same thing in javascript. I'll use python simply to see if I can.

````python
import json 
import base64
from Crypto.Cipher import AES

# provided by: https://github.com/KayDeeTee/Hollow-Knight-SaveManager/
c_sharp_header = bytes.fromhex('0001000000FFFFFFFF01000000000000000601000000')
end_header = bytes([11])
aes_key = b'UKu52ePUBwetZ9wNX88o54dnfKRu0T1l'


def read_binary_file(file_path: str):
    with open(file_path, 'rb') as binary_file:
        data = binary_file.read()
    return data


def read_json_file(file_path: str):
    with open(file_path, 'r') as text_file:
        data = json.load(text_file)
    return data

def create_cipher():
    return AES.new(aes_key, AES.MODE_ECB)


def decode_and_decrypt(bytes_):
    # remove header and last byte
    bytes_ = bytes_[len(c_sharp_header): len(bytes_) - 1]
    # decode the base64 bytes
    bytes_ = base64.b64decode(bytes_)
    # create a aes ecb-mode cipher
    cipher = create_cipher()
    # decrypt
    bytes_ = cipher.decrypt(bytes_)
    # finally remove padding
    return bytes_[:-bytes_[-1]]







def decrypt_handle(data_file):
    with open('user1.dat', 'rb') as binary_file:
        data_as_bytes = binary_file.read()        
    data_as_bytes = read_binary_file(data_file)
    data_as_bytes = decode_and_decrypt(data_as_bytes)
    print('ok')
    print(f'writing to file: test2.json)')
    dict_ = json.loads(data_as_bytes)
    str_to_write = json.dumps(dict_, indent=2, sort_keys=False)
    return json.loads(str_to_write)



    
    
    
````

````python
json_data = decrypt_handle('user1.dat')
````
